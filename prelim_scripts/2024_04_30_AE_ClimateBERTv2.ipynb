```python
import pandas as pd
from transformers import AutoModelForSequenceClassification, AutoTokenizer, pipeline
from transformers.pipelines.pt_utils import KeyDataset
import datasets
from tqdm.auto import tqdm

dataset_name = "climatebert/climate_specificity"
model_name = "climatebert/distilroberta-base-climate-specificity"

# If you want to use your own data, simply load them as 🤗 Datasets dataset, see https://huggingface.co/docs/datasets/loading
dataset = datasets.load_dataset("csv", data_files="/Users/aelfarsdottir/Documents/Stanford/PhD/ESGs/Brownspinning/Assets/preprocessed_data/initiative_details_JUNE.csv", split = "train")
# how do I format my dataset as this text: label: ? 
# just had to take out the NA's
# > write.csv(data.frame(text = namematched_all_merged_fill_vars$initiative_details[!(is.na(namematched_all_merged_fill_vars$initiative_details)) & nchar(namematched_all_merged_fill_vars$initiative_details) > 0]), "preprocessed_data/initiative_details_APRIL.csv", row.names=FALSE)

model = AutoModelForSequenceClassification.from_pretrained(model_name)
tokenizer = AutoTokenizer.from_pretrained(model_name, max_len=512)

pipe = pipeline("text-classification", model=model, tokenizer=tokenizer, device=0)

# See https://huggingface.co/docs/transformers/main_classes/pipelines#transformers.pipeline
# for out in tqdm(pipe(KeyDataset(dataset, "text"), padding=True, truncation=True)):
#    print(out)

# thank you GPT3.5! for saving scores and labels as dataset columns
# Initialize lists to store predicted labels and scores
all_text = []
predicted_labels = []
scores = []

# Initialize counter
counter = 0

try:
    # Classify text and collect predicted labels and scores
    # Classify text with dynamic padding
    for example in tqdm(KeyDataset(dataset, "text")):
        counter += 1
        all_text.append(example)
        output = pipe(example, padding=True, truncation=True)
        predicted_label = output[0]['label']
        predicted_score = output[0]['score']
        predicted_labels.append(predicted_label)
        scores.append(predicted_score)
        if counter % 10 == 0:
            print(f"Processed {counter} entries, label: {predicted_label}, score: {predicted_score}")
       # if counter == 10170:
       #     raise StopIteration  # Break out of the loop when reaching 10890 entries
except StopIteration:
    pass  # Do nothing, just continue after breaking the loop

# Create a DataFrame from the collected predictions
df = pd.DataFrame({
    'text': all_text,
    'predicted_label': predicted_labels,
    'score': scores
})

# Save the updated dataset to your folder
df.to_csv('/Users/aelfarsdottir/Documents/Stanford/PhD/ESGs/Brownspinning/Assets/preprocessed_data/initiative_details_JUNE_classified.csv')


# # Use tqdm to wrap the pipe function call for progress updates
# for out in tqdm(pipe(KeyDataset(dataset, "text"), padding = True, truncation = True)):
    
#     for prediction in out: 
        
#         # Access the output 'out' and collect predicted label and score
#         label = prediction['label']
#         score = prediction['score']

#         # Append label and score to respective lists
#         predicted_labels.append(label)
#         scores.append(score)
```
