```python
import polars as pl
```

```python
df = pl.DataFrame({
    'column': ['a', 'b', 'c', 'c', 'e', 'd', 'd', 'e', 'e']
})
```

```python
df = (
    df
    .group_by('column')
    .agg(pl.count().alias('number'))
    .with_columns(pl.int_ranges(1, pl.col('number') + 1).alias('number'))
    .explode('number')
)
```
